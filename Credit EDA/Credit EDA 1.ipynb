## Importing Libraries

```python
# import libraries
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline

pd.set_option('display.max_columns',None)
pd.set_option('display.max_rows',200)
```

## Loading Data

```python
# load application_data file
application_data = pd.read_csv("application_data.csv")
application_data.head()
```

### Data Overview

```python
# Basic overview
print("Shape:", application_data.shape)
display(application_data.head())
print("\nColumns:", application_data.columns.tolist())
```

```python
# Info & summary stats
application_data.info()
display(application_data.describe().T)
```

```python
# Missing values table (counts + percent)
missing = application_data.isnull().sum().sort_values(ascending=False)
missing_pct = (missing / len(application_data) * 100).round(2)
missing_df = pd.DataFrame({'missing_count': missing, 'missing_pct': missing_pct})
display(missing_df[missing_df['missing_count'] > 0].head(80))
```

```python
# Small missing data heatmap (shows patterns)
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12,6))
sns.heatmap(application_data.isnull(), cbar=False)
plt.title('Missing values heatmap (True = missing)')
plt.show()
```

```python
# Duplicates
dups = application_data.duplicated().sum()
print("Duplicate rows:", dups)
# Optionally inspect duplicates
if dups:
    display(application_data[application_data.duplicated(keep=False)].head(10))
```

```python
# Look for suspicious values in numeric columns
suspicious = [365243, 999999, 0]  # add others if you know them
for c in application_data.select_dtypes(include=['int64','float64']).columns:
    top = application_data[c].value_counts(dropna=False).head(6)
    if any(v in top.index for v in suspicious):
        print(f"\nColumn {c} top values:")
        display(top)
```

```python
application_data['DAYS_EMPLOYED'].replace({365243: np.nan}, inplace=True)
```

```python
# Show dtype counts
print(application_data.dtypes.value_counts())

# Convert some logical columns to category (example)
for col in ['FLAG_OWN_CAR', 'FLAG_OWN_REALTY', 'CODE_GENDER']:
    if col in application_data.columns:
        application_data[col] = application_data[col].astype('category')

# Example: if DAYS_BIRTH is negative (days before application), keep as numeric but convert to years later
```

```python
# Identify numeric and categorical
num_cols = application_data.select_dtypes(include=['number']).columns.tolist()
cat_cols = application_data.select_dtypes(include=['object','category']).columns.tolist()

# Numeric: median imputation + flag
for c in num_cols:
    if application_data[c].isnull().sum() > 0:
        application_data[c + '_missing_flag'] = application_data[c].isnull().astype(int)
        med = application_data[c].median()
        application_data[c].fillna(med, inplace=True)

# Categorical: mode imputation + flag
for c in cat_cols:
    if application_data[c].isnull().sum() > 0:
        application_data[c + '_missing_flag'] = application_data[c].isnull().astype(int)
        mode = application_data[c].mode().iloc[0]
        application_data[c].fillna(mode, inplace=True)
```

```python
if 'TARGET' in application_data.columns:
    print("Target value counts:")
    display(application_data['TARGET'].value_counts())
    print("\nTarget proportions:")
    display(application_data['TARGET'].value_counts(normalize=True).round(3))
else:
    print("TARGET column not found. Check dataset for target column name.")
```

```python
# Debt-to-income ratio (AMT_ANNUITY is monthly/yearly? use dataset docs — this is a simple formula)
if set(['AMT_ANNUITY','AMT_INCOME_TOTAL']).issubset(application_data.columns):
    application_data['debt_to_income'] = application_data['AMT_ANNUITY'] / (application_data['AMT_INCOME_TOTAL'] + 1e-9)

# Credit term (how many months/periods to repay) — careful if AMT_ANNUITY 0 or missing
if set(['AMT_CREDIT','AMT_ANNUITY']).issubset(application_data.columns):
    application_data['credit_term'] = application_data['AMT_CREDIT'] / (application_data['AMT_ANNUITY'] + 1e-9)

# Income per family member
if 'CNT_FAM_MEMBERS' in application_data.columns and 'AMT_INCOME_TOTAL' in application_data.columns:
    application_data['income_per_person'] = application_data['AMT_INCOME_TOTAL'] / application_data['CNT_FAM_MEMBERS'].replace(0,1)
```

```python
import matplotlib.pyplot as plt
application_data['TARGET'].value_counts().plot(kind='bar')
plt.title('Target counts (0 = repaid, 1 = default)')
plt.show()
```

```python
if 'NAME_CONTRACT_TYPE' in application_data.columns:
    tmp = application_data.groupby('NAME_CONTRACT_TYPE')['TARGET'].mean().sort_values()
    tmp.plot(kind='bar', figsize=(6,3))
    plt.ylabel('Default rate')
    plt.title('Default rate by contract type')
    plt.show()
```

```python
# Log transform (keep original)
if 'AMT_INCOME_TOTAL' in application_data.columns:
    application_data['AMT_INCOME_TOTAL_log'] = np.log1p(application_data['AMT_INCOME_TOTAL'].clip(lower=0))
```

```python
application_data.to_csv("application_data_cleaned.csv", index=False)
```
